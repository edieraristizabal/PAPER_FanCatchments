```python
cuca_CHIRPS = pd.read_csv('G:\Mi unidad\INVESTIGACION\SEGUIMIENTO ESTUDIANTES\PAE\Emanuel Castillo\cucaracho.csv')
gerona_CHIRPS = pd.read_csv('G:\Mi unidad\INVESTIGACION\SEGUIMIENTO ESTUDIANTES\PAE\Emanuel Castillo\Gerona.csv')
palmas_CHIRPS = pd.read_csv('G:\Mi unidad\INVESTIGACION\SEGUIMIENTO ESTUDIANTES\PAE\Emanuel Castillo\palmas.csv')
caldas_CHIRPS = pd.read_csv('G:\Mi unidad\INVESTIGACION\SEGUIMIENTO ESTUDIANTES\PAE\Emanuel Castillo\Caldas.csv')
```

```python
gerona_CHIRPS=gerona_CHIRPS['value']
caldas_CHIRPS=caldas_CHIRPS['value']
cuca_CHIRPS=cuca_CHIRPS['value']
palmas_CHIRPS=palmas_CHIRPS['value']
```

```python
daily=pd.date_range('1981-01-01','2016-12-30', freq='D')
print(len(daily))
```

```python
gerona_CHIRPS = pd.Series(data=gerona_CHIRPS.values, index=daily)
cuca_CHIRPS = pd.Series(data=cuca_CHIRPS.values, index=daily)
palmas_CHIRPS = pd.Series(data=palmas_CHIRPS.values, index=daily)
caldas_CHIRPS = pd.Series(data=caldas_CHIRPS.values, index=daily)
```

```python
dgerona_EPM[dgerona_EPM<0]=0
dcuca_EPM[dcuca_EPM<0]=0
dpalmas_EPM[dpalmas_EPM<0]=0
dcaldas_EPM[dcaldas_EPM<0]=0
```

```python
dgerona_EPM.plot()
```

```python
dgerona_EPM=dgerona_EPM[dgerona_EPM.index.year>1995]
dcuca_EPM=dcuca_EPM[dcuca_EPM.index.year>1990]
dpalmas_EPM=dpalmas_EPM[dpalmas_EPM.index.year>1980]
dcaldas_EPM=dcaldas_EPM[dcaldas_EPM.index.year>1980]
```

```python
print(len(dgerona_EPM))
print(len(gerona_CHIRPS))
```

```python
dgerona_EPM.drop(dgerona_EPM.index[-1],axis=0,inplace=True)
dcuca_EPM.drop(dcuca_EPM.index[-1],axis=0,inplace=True)
dpalmas_EPM.drop(dpalmas_EPM.index[-1],axis=0,inplace=True)
dcaldas_EPM.drop(dcaldas_EPM.index[-1],axis=0,inplace=True)
```

```python
gerona_CHIRPS=gerona_CHIRPS[gerona_CHIRPS.index.year>1995]
cuca_CHIRPS=cuca_CHIRPS[cuca_CHIRPS.index.year>1990]
palmas_CHIRPS=palmas_CHIRPS[palmas_CHIRPS.index.year>1980]
caldas_CHIRPS=caldas_CHIRPS[caldas_CHIRPS.index.year>1980]
```

```python
print(len(gerona_CHIRPS))
print(len(dgerona_EPM))
```

# Correlación Diaria

## Estación Gerona

```python
from sklearn.metrics import r2_score
print(r2_score(gerona_CHIRPS,dgerona_EPM))
print(np.corrcoef(gerona_CHIRPS,dgerona_EPM))
```

```python
plt.scatter(gerona_CHIRPS,dgerona_EPM,marker='.',c='black',s=12)
plt.plot([0,100],[0,100],linestyle='--',color='k')
plt.xlabel('CHIRPS')
plt.xlim(0,100)
plt.ylabel('Gerona Rain gauge')
plt.ylim(0,100)
plt.text(75,90,'Corr=0.307',bbox=dict(boxstyle='round',facecolor='white'))
plt.title('Daily (1995-2016)')
plt.grid(True)
plt.savefig('G:\Mi unidad\INVESTIGACION\SEGUIMIENTO ESTUDIANTES\PAE\Emanuel Castillo/dgerona_corr.tiff');
```
