# CORRELATION

```python
cor = gdffloat.corr(method ='pearson')
cor_recortado=cor.iloc[14:,0:14]
from mpl_toolkits.axes_grid1 import make_axes_locatable
fig, ax = plt.subplots(figsize=(10,10))
im = ax.imshow(cor_recortado.T, cmap='YlGn')
# We want to show all ticks...
ax.set_yticks(np.arange(len(cor_recortado.columns)))
ax.set_xticks(np.arange(len(cor_recortado.index)))
# ... and label them with the respective list entries
ax.set_yticklabels(cor_recortado.columns)
ax.set_xticklabels(cor_recortado.index)
plt.setp(ax.get_xticklabels(), rotation=90, ha="right",rotation_mode="anchor")
# Create colorbar
divider = make_axes_locatable(ax)
cax = divider.append_axes("right", size="5%", pad=0.2)
plt.colorbar(im, cax=cax)
plt.savefig('G:\My Drive\INVESTIGACION\PAPERS\ELABORACION\PAPER_FanCatchments\FIGURES\corre.png');
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming debris_fan_data and catchments_data_corrected are pre-defined DataFrames with the correct columns

# Ensure the columns are lists
debris_fan_vars = list(fanfloat.columns)
catchments_vars = list(catfloat.columns)

# Manually compute pairwise Spearman correlation between debris fan and catchments variables
spearman_correlation_matrix = pd.DataFrame(index=debris_fan_vars, columns=catchments_vars)

for fan_var in debris_fan_vars:
    for catch_var in catchments_vars:
        # Ensure we can safely calculate correlations with non-empty data
        if fanfloat[fan_var].notnull().all() and catfloat[catch_var].notnull().all():
            spearman_correlation_matrix.loc[fan_var, catch_var] = fanfloat[fan_var].corr(
                catfloat[catch_var], method='spearman')

# Convert the result to numeric for better visualization
spearman_correlation_matrix = spearman_correlation_matrix.apply(pd.to_numeric, errors='coerce')

# Plotting the heatmap for the Spearman correlation matrix between debris fan and catchments variables
plt.figure(figsize=(14, 14))  # Adjust figure size to maintain square blocks
sns.heatmap(spearman_correlation_matrix, cmap='coolwarm', annot=False, vmin=-1, vmax=1, square=True, 
            cbar_kws={"shrink": 0.35, "ticks": [-1, -0.5, 0, 0.5, 1]})  # Adjusting the color bar size
plt.xticks(rotation=90)  # Rotate x-axis labels for better readability
plt.yticks(rotation=0)  # Keep y-axis labels horizontal
plt.show()
```

## Boxplots

```python
red_circle = dict(markerfacecolor='red', marker='o', markeredgecolor='white')
fig, axs = plt.subplots(1, len(catfloat.columns), figsize=(60,10))
for i, ax in enumerate(axs.flat):
    ax.boxplot(catfloat.iloc[:,i], flierprops=red_circle)
    ax.tick_params(axis='y', labelsize=14)
    ax.set_xticks([])
    ax.set_xlabel(catfloat.columns[i], fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig("G:\My Drive\INVESTIGACION\PAPERS\ELABORACION\PAPER_FanCatchments/Figures/catchments_boxplots.png")
```

```python
# Creating the scatter plot with marker color based on the variable $Sn_f$
plt.figure(figsize=(8, 6))
scatter = plt.scatter(gdf['$A_c$'], gdf['$A_f$'], 
                      s=gdf['$R_c$']*10, alpha=0.7, marker='o', 
                      c=gdf['$Sn_f$'], cmap='viridis')

# Adding a custom legend for marker sizes and renaming labels to "10km" and "100km", removing 50
handles, labels = scatter.legend_elements(prop="sizes", num=[10, 100], alpha=0.6)
new_labels = ['10km', '100km']  # Custom labels
plt.legend(handles, new_labels, title="Catchment relief", loc="upper left", frameon=True, 
           borderpad=0.5, labelspacing=1, title_fontproperties={'weight': 'bold', 'size': 13},numpoints=1)

# Applying log scale and updating axis labels
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Catchment area ($km^2$)', fontsize=14, fontweight='bold')
plt.ylabel('Debris fan area ($km^2$)', fontsize=14, fontweight='bold')

# Adding a colorbar for Sn_f with the degree symbol (°)
cbar = plt.colorbar(scatter)
cbar.set_label('Fan Slope (°)', fontsize=13, fontweight='bold')

# Display axis on top and right
plt.gca().spines['top'].set_visible(True)
plt.gca().spines['right'].set_visible(True)

plt.show()
```
