```
import seaborn as sns
plt.figure(figsize=(30,30))
cor = np.round(data.iloc[:,:-1].corr(method ='pearson'),4)
cor = data.corr(method ='pearson')
plt.title('Análisis de correlación')
mask=np.array(cor)
mask[np.tril_indices_from(mask)]=False
sns.heatmap(cor, annot=True, mask=mask, cmap=plt.cm.Reds, annot_kws={"size":12})
sns.set(font_scale=2)
plt.tight_layout()

resultados = '/content/drive/MyDrive/TESIS Emanuel - Ing. Geológica 2020/Imágenes de Resultados '
plt.savefig(resultados+'/Corrpearsontot1.png')
```

```
def generate_accuracy_and_heatmap(model, x, y):
#     cm = confusion_matrix(y,model.predict(x))
#     sns.heatmap(cm,annot=True,fmt="d")
    ac = accuracy_score(y,model.predict(x))
    f_score = f1_score(y,model.predict(x))
    print('Accuracy is: ', ac)
    print('F1 score is: ', f_score)
    print ("\n")
    print (pd.crosstab(pd.Series(model.predict(x), name='Predicted'),
                       pd.Series(y['Tipo de Cuenca'],name='Actual')))
    return 1
```

```
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import validation_curve
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, Normalizer
from sklearn.model_selection import train_test_split
from sklearn.model_selection import KFold
import random 
from sklearn.model_selection import KFold
from sklearn.model_selection import learning_curve
import seaborn as sns
from sklearn.cluster import KMeans, SpectralClustering
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neighbors import KNeighborsClassifier
from yellowbrick.contrib.classifier import DecisionViz
from sklearn.model_selection import ShuffleSplit
from sklearn.metrics import recall_score
from sklearn.metrics import fbeta_score, make_scorer
from sklearn.svm import SVC
```

```
recall = make_scorer(recall_score)
```

```
X_fs = data.loc[:, data.columns != target]
Y_fs = data.loc[:, data.columns == target]
```

```
X_train_fs, X_test_fs, Y_train_fs, Y_test_fs = train_test_split(X_fs,Y,test_size=0.20, random_state=0)
```

##Standard sacaler

```
StandardScaler=StandardScaler()
StandardScaler.fit(X_fs) #Compute the mean and std
X_fs_SS=StandardScaler.transform(X_fs) #Perform standardization by centering and scaling
X_fs_SS= pd.DataFrame(X_fs_SS, columns=X_fs.columns)

X_fs_SS.head(6)
```

##REGRESIÓN LOGISTICA (LR) -   Linear supervised parametric algorithms

###validation curve

```
lr= LogisticRegression(random_state=0) #COLOCAR RANDOM_STATE!
```

```
C_SS= np.linspace(0.1,0.8, 20)
kfold=KFold(n_splits=5) #kfold NO revuelve
lr_train_scores_fs_SS, lr_val_scores_fs_SS = validation_curve(lr,X_train_fs, Y_train_fs, param_name='C', param_range=C_SS, cv=kfold,scoring=recall)
```

```
lr_train_sizes_fs_SS=C_SS
lr_train_scores_mean_fs_SS=np.mean(lr_train_scores_fs_SS,axis=1)
lr_train_scores_std_fs_SS=np.std(lr_train_scores_fs_SS,axis=1)
lr_val_scores_mean_fs_SS=np.mean(lr_val_scores_fs_SS,axis=1)
lr_val_scores_std_fs_SS=np.std(lr_val_scores_fs_SS,axis=1)
```

```
plt.figure()
plt.title("Validation Curves (Lr_fs_SS)\n")    
plt.xlabel("C")
plt.ylabel("Score: Recall")
# plt.fill_between(lr_train_sizes_fs_SS, lr_train_scores_mean_fs_SS - lr_train_scores_std_fs_SS, lr_train_scores_mean_fs_SS + lr_train_scores_std_fs_SS, alpha=0.1,color="darkorange")
# plt.fill_between(lr_train_sizes_fs_SS, lr_val_scores_mean_fs_SS - lr_val_scores_std_fs_SS,lr_val_scores_mean_fs_SS + lr_val_scores_std_fs_SS, alpha=0.1, color="navy")
plt.plot(lr_train_sizes_fs_SS, lr_train_scores_mean_fs_SS, 'o-', color="darkorange", label="Training score")
plt.plot(lr_train_sizes_fs_SS, lr_val_scores_mean_fs_SS, 'o-', color="navy", label="Cross-validation score")
plt.legend(loc="best")
```

###Grid search

```
c_random_SS=np.random.uniform(0.1,0.8,5)
c_random_SS
```

```
c_random_SS=[0.2747268 , 0.36377983, 0.18847497, 0.57173582, 0.31255873]
```

```
lr_parameters_fs_SS = {'C':c_random_SS, 'class_weight':('balanced',None),'solver':('newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga')}
lr_model_fs_SS = lr
lr_grid_fs_SS= GridSearchCV(estimator=lr_model_fs_SS, param_grid=lr_parameters_fs_SS,scoring=recall)
lr_grid_fs_SS.fit(X_train_fs, Y_train_fs)
print(lr_grid_fs_SS.best_score_)
print(lr_grid_fs_SS.best_estimator_)
```

###Learning curve

```
#C=0.10595623787796078, score:0.7601470068646938
lr_fs_SS_best=lr_grid_fs_SS.best_estimator_
kfold = KFold(n_splits=5)
lr_lc_fs_SS = learning_curve(lr_fs_SS_best,X_train_fs, Y_train_fs,cv=kfold,scoring=recall)
```

```
lr_lc_train_sizes_fs_SS=lr_lc_fs_SS[0]
lr_lc_train_scores_mean_fs_SS=np.mean(lr_lc_fs_SS[1],axis=1)
lr_lc_train_scores_std_fs_SS=np.std(lr_lc_fs_SS[1],axis=1)
lr_lc_validation_scores_mean_fs_SS=np.mean(lr_lc_fs_SS[2],axis=1)
lr_lc_validation_scores_std_fs_SS=np.std(lr_lc_fs_SS[2],axis=1)
```

```
def generate_accuracy_and_heatmap(model, x, y):
#     cm = confusion_matrix(y,model.predict(x))
#     sns.heatmap(cm,annot=True,fmt="d")
    ac = accuracy_score(y,model.predict(x))
    f_score = f1_score(y,model.predict(x))
    print('Accuracy is: ', ac)
    print('F1 score is: ', f_score)
    print ("\n")
    print (pd.crosstab(pd.Series(model.predict(x), name='Predicted'),
                       pd.Series(y['Torrential flow reports'],name='Actual')))
    return 1
```

```
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
from sklearn.decomposition import PCA
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
```

```
X=pd.read_excel('/content/drive/My Drive/TESIS Emanuel - Ing. Geológica/ML4.xlsx')
```

```
X.head()
```

```
y=X['Torrential flow reports']
X.drop(['Torrential flow reports','Fan perimeter (km)', ' Stream length ratio', ' Ro Coefficient','Fan length (km)','Form factor of fan','Fan relief (km)','Fan ruggednes ','Main stream order','Number of streams ',' Stream length (km)','Main stream length (km)','Bifurcation ratio','Catchment area (km^2)','Catchment perimeter (km)',' Elongation ratio','Circularity ratio','Compactness coefficien','Texture ratio','Fitness ratio','Drainage intensity','Infiltration numbe','Torrentiality coefficient','Length of overland flow','Catchment relief (km)','Mean basin slope (°)','Relief ratio','Ruggedness number','Melton index'], axis=1, inplace=True)
X.head()
```

```
scaler=StandardScaler()
scaler.fit(X)
X_scaled=scaler.transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)
```

```
n = len(X_scaled.columns)
pca = PCA(n_components=n).fit(X_scaled)
pca_samples = pca.transform(X_scaled)
```

```
plt.plot(pca.explained_variance_ratio_)
plt.xlabel('Number of components')
plt.ylabel('Explained variance')
plt.show()
```

```
plt.plot(np.cumsum(pca.explained_variance_ratio_))
plt.xlabel('number of components')
plt.ylabel('cumulative explained variance')
plt.show()
```

```
xvector = pca.components_[0] 
yvector = pca.components_[1]

xs = pca.transform(X_scaled)[:,0] 
ys = pca.transform(X_scaled)[:,1]

mask1=np.ma.masked_where(y < 1,xs )
mask2=np.ma.masked_where(y < 1,ys ) 

## Para visualizar las proyecciones de cada variable en los componentes se utiliza la siguiente función
for i in range(len(xvector)):
# arrows project features (ie columns from csv) as vectors onto PC axes
    plt.arrow(0, 0, xvector[i]*max(xs), yvector[i]*max(ys),
              color='r', width=0.0005, head_width=0.0025)
    plt.text(xvector[i]*max(xs)*1.2, yvector[i]*max(ys)*1.2,
             list(X_scaled.columns.values)[i], color='r') 

plt.scatter(xs, ys, s=70,marker='x',c='blue', label='MenM')
plt.scatter(mask1,mask2,facecolors='black', edgecolors='black', s=70, alpha=0.5, label='No MenM')
plt.tick_params('y', colors='k', labelsize=12, length=2)
plt.tick_params('x', colors='k', labelsize= 12, length=2)
plt.xlabel("Componente Principal 1", fontsize=16)
plt.ylabel("Componente Principal 2", fontsize=16)
plt.legend(fontsize=14)
plt.show()
```

```
def pca_results(X_scaled, pca):
    
    dimensions = ['Dim. {}'.format(i) for i in range(1,len(pca.components_)+1)]
    
    components = pd.DataFrame(np.round(pca.components_, 4), columns = X_scaled.keys()) 
    components.index = dimensions

    ratios = pca.explained_variance_ratio_.reshape(len(pca.components_), 1) 
    variance_ratios = pd.DataFrame(np.round(ratios, 4), columns = ['Explained Variance']) 
    variance_ratios.index = dimensions

    fig, ax = plt.subplots(figsize = (20,13))

    components.plot(ax = ax, kind = 'bar')
    ax.set_ylabel("Feature Weights") 
    ax.set_xticklabels(dimensions, rotation=0)

    for i, ev in enumerate(pca.explained_variance_ratio_): 
        ax.text(i-0.40, ax.get_ylim()[1] + 0.05, "Expl. Var.\n %.4f"%(ev))

    return pd.concat([variance_ratios, components], axis = 1)

pca_results = pca_results(X_scaled, pca)
```
