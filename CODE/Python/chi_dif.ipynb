```python
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
data=gpd.read_file('/content/drive/MyDrive/INVESTIGACION/POSDOC/Paper1/chi_dif_recents_slope.shp')
data.head()
```

```python
fig, ax = plt.subplots(figsize = (15,5))
plt.scatter(data.chi_range,data.Name_count, c=data._mean)
ax.set_xlim(0,9)
ax.set_ylim(0,80)
plt.xlabel("$\chi$ difference",fontsize=14)
plt.ylabel("Landslides number",fontsize=14)
clb=plt.colorbar()
clb.ax.tick_params(labelsize=8) 
clb.set_label('Mean slope',fontsize=14,rotation=270,labelpad=15)
#plt.savefig('/content/drive/MyDrive/INVESTIGACION/POSDOC/Paper1/chi_diff_slope.png',dpi=200,transparent=True)
```
