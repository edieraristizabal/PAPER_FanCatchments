```python
cor = np.round(gdf_cat.iloc[:,:-1].corr(method ='pearson'),4)
cor = gdf_cat.corr(method ='pearson')
cor_recortado=cor.iloc[15:,1:15]
from mpl_toolkits.axes_grid1 import make_axes_locatable
fig, ax = plt.subplots(figsize=(10,10))
im = ax.imshow(cor_recortado.T, cmap='YlGn')
# We want to show all ticks...
ax.set_yticks(np.arange(len(cor_recortado.columns)))
ax.set_xticks(np.arange(len(cor_recortado.index)))
# ... and label them with the respective list entries
ax.set_yticklabels(cor_recortado.columns)
ax.set_xticklabels(cor_recortado.index)
plt.setp(ax.get_xticklabels(), rotation=90, ha="right",rotation_mode="anchor")
# Create colorbar
divider = make_axes_locatable(ax)
cax = divider.append_axes("right", size="5%", pad=0.2)
plt.colorbar(im, cax=cax)
plt.savefig('/content/drive/MyDrive/INVESTIGACION/POSDOC/Paper1.2/corre.png',dpi=200,transparent=True)
```

```python
sample=fan["Fan area (km2)"]
h,e = np.histogram(sample, bins=100, density=True)
x = np.linspace(e.min(), e.max())
# plot the histogram
plt.figure(figsize=(4,2))
plt.bar(e[:-1], h, width=np.diff(e), ec='k', align='edge', label='histogram')
# plot the real KDE
kde = sts.gaussian_kde(sample)
plt.plot(x, kde.pdf(x), c='C1', lw=2, label='KDE')
```
