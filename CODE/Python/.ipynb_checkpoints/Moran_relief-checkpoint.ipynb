```python
%matplotlib inline
import numpy as np
import pandas as pd
import geopandas as gpd
import seaborn as sbn
import matplotlib
import matplotlib.pyplot as plt
from matplotlib_scalebar.scalebar import ScaleBar
import contextily as ctx
import pysal
from pysal.explore import esda
from pysal.viz import splot
from pysal.lib import weights
from splot.esda import lisa_cluster
from splot.esda import plot_moran
from splot.esda import plot_local_autocorrelation, moran_scatterplot

import warnings
warnings.filterwarnings("ignore")
```

### Catchments

```python
catchments=gpd.read_file("G:/My Drive/INVESTIGACION/POSDOC/Data/Vector/hypso3_rainfall.gpkg")
catchments.head(2)
```

Se eliminan columnas que no se van a utilizar, se proyecta a coordenadas planas, y se crea uan columna con el index para que cada cuenca tenga un valor unico. El archivo original repetia numeros de cuencas tanto en la columna llamada 'Id' como en al llamada 'gridcode'

```python
import uuid
cat=catchments.drop(['Id','gridcode','Área','area_curva','_count','_sum','_mean','_stdev','_max','_range','_variance'],axis=1)
cat=cat.to_crs(epsg=32618)
cat['id'] = [uuid.uuid4().hex.upper()[0:5] for _ in range(len(cat.index))]
cat.head(2)
```

### Recent landslides

Se importan lo movimientos en masa recientes que incluye el valor de lluvia media anual en la columna 'SAMPLE_1', el cual se elaboro en QGIS.

```python
recent=gpd.read_file("G:/My Drive/INVESTIGACION/POSDOC/Data/Vector/recients.shp")
recent.head(2)
```

Se eliminan las columnas que nos e van a utilizar y se cambia el nombre de la columna que tiene la lluvia. Y se proyecta a coordenadas planas.

```python
rec=recent.drop(["Name","descriptio","timestamp","begin","end","altitudeMo","tessellate","extrude","visibility","drawOrder","icon","Fecha"],axis=1)
rec.rename(columns={"SAMPLE_1":"rainfall"}, inplace=True)
rec=rec.to_crs(epsg=32618)
rec.plot()
rec.head(2)
```

Se realiza una interseccion espacial entre las cuencas 'cat' y los movimientos en masa recientes 'rec'

```python
rec_cat = gpd.sjoin(rec, cat, how="inner")
rec_cat.head(2)
```

se agrupa por el numero de la cuenca para saber cuantos movimientso en masa hay por cuenca y se crea una dataframe.

```python
rec_by_cat=rec_cat.groupby("id").size()
land_cat=pd.DataFrame({"rec_count": rec_by_cat})
land_cat.head(2)
```

```python
cat = cat.join(land_cat,on="id")
cat.head(2)
```

```python
cat.set_index('id',drop=False,inplace=True)
cat["rec_count"]=cat["rec_count"].fillna(0)
cat.head(2)
```

```python
f, ax = plt.subplots(1, figsize=(5, 5))
cat.plot(column='hypso_inte', scheme='Quantiles', legend=True, ax=ax)
ax.set_axis_off()
f.set_facecolor('0.75')
f.suptitle('Hypsometric index', size=30)
plt.show()
```

```python
w_queen=weights.Queen.from_dataframe(cat,idVariable="id")
w_queen
```

```python
queen_card = pd.Series(w_queen.cardinalities)
sbn.distplot(queen_card, bins=10);
```

```python
ax = cat.plot(color='k', figsize=(7, 7))
cat.loc[w_queen.islands, :].plot(color='red', ax=ax);
```

```python
w_queen.transform = 'R'
```

```python
w_dist30km = weights.DistanceBand.from_dataframe(cat, 30000, binary=False)
```

```python
f, ax = plt.subplots(1, figsize=(10, 10))
cat.plot(ax=ax, facecolor='k', linewidth=0.1)
focus = cat.loc[['048F3'], ['geometry']]
focus.plot(facecolor='red', alpha=1, linewidth=0, ax=ax)
#neis = cat.loc[w_dist30km['048F3'], :]
#neis.plot(ax=ax, facecolor='lime', linewidth=0)
buf = focus.centroid.buffer(30000)
buf.plot(edgecolor='red', facecolor='none', ax=ax, linewidth=4)
pts = np.array([(pt.x, pt.y) for pt in cat.centroid])
ax.plot(pts[:, 0], pts[:, 1], color='#00d8ea',linewidth=0, alpha=0.75, marker='o', markersize=4)
f.suptitle("Neighbors within 30km of `048F3`")
plt.show()
```

```python
cat['w_hypso'] = weights.lag_spatial(w_queen, cat['hypso_inte'])
```

```python
mi = esda.Moran(cat['hypso_inte'], w_dist30km)
plot_moran(mi);
```

```python
sbn.distplot(lisa.Is)
```

```python
f, ax = plt.subplots(1, figsize=(6,6))
cat['Is'] = lisa.Is
cat.plot(column='Is', cmap='viridis', scheme='quantiles',k=10, edgecolor='white', linewidth=0.1, alpha=0.75, legend=True,ax=ax);
ax.set_axis_off()
```

```python
from matplotlib.colors import ListedColormap
f, axs = plt.subplots(nrows=2, ncols=2, figsize=(9, 9))
axs = axs.flatten()

# Subplot 1 #
ax = axs[0]
cat.plot(column='Is', cmap='viridis', scheme='quantiles', k=10, edgecolor='white', linewidth=0.1, alpha=0.75, legend=True, ax=ax)
ax.set_aspect('equal')
ax.set_axis_off()

# Subplot 2 #
ax = axs[1]
q_labels = ['Q1', 'Q2', 'Q3', 'Q4']
labels = [q_labels[i-1] for i in lisa.q]
hmap = ListedColormap([ 'red', 'lightblue', 'blue', 'pink'])
cat.assign(cl=labels).plot(column='cl', categorical=True,k=2, cmap=hmap, linewidth=0.1, ax=ax,edgecolor='white', legend=True)
ax.set_aspect('equal')
ax.set_axis_off()

# Subplot 3 #
ax = axs[2]
sig = 1 * (lisa.p_sim < 0.05)
hmap = ListedColormap(['grey','black'])
labels = ['non-sig.', 'significant'] 
labels = [labels[i] for i in sig]
cat.assign(cl=labels).plot(column='cl', categorical=True,k=2, cmap=hmap, linewidth=0.1, ax=ax,edgecolor='white', legend=True)
ax.set_aspect('equal')
ax.set_axis_off()
                            
# Subplot 4 #
ax = axs[3]
hotspot = 1 * (sig * lisa.q==1)
coldspot = 3 * (sig * lisa.q==3)
doughnut = 2 * (sig * lisa.q==2)
diamond = 4 * (sig * lisa.q==4)
spots = hotspot + coldspot + doughnut + diamond
spot_labels = [ '0 ns', '1 hot spot', '2 doughnut', '3 cold spot', '4 diamond']
labels = [spot_labels[i] for i in spots]
hmap = ListedColormap([ 'grey', 'red', 'lightblue', 'blue', 'pink'])
cat.assign(cl=labels).plot(column='cl', categorical=True,k=2, cmap=hmap, linewidth=0.1, ax=ax,edgecolor='white', legend=True)
ax.set_aspect('equal')
ax.set_axis_off()

plt.show()
```
