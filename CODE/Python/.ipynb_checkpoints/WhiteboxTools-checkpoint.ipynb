## Data Tools
https://www.whiteboxgeo.com/manual/wbt_book/available_tools/data_tools.html

```python
# hillshade
wbt.hillshade(
    "sucio.tif", 
    "sucio_hillshade.tif", 
    azimuth=315.0, 
    altitude=30.0, 
    zfactor=None, 
)
```

```python
# slope
wbt.slope(
    'arma.tif', 
    'arma_slope.tif', 
    zfactor=None, 
    units="degrees", # 'degrees', 'radians', 'percent'
)
```

```python
#aspect
wbt.aspect(
    'porce.tif', 
    'porce_aspect.tif', 
    zfactor=None, 
)
```

```python
# plan curvature
wbt.plan_curvature(
    'porce.tif', 
    'porce_curvature_plan.tif', 
    log=False, #Display output values using a log-scale
    zfactor=None, 
)
```

```python
#profile curvature
wbt.profile_curvature(
    'porce.tif', 
    'porce_curvature_profile.tif', 
    log=False, 
    zfactor=None, 
)
```

```python
# total curvature Total curvature measures the curvature of the topographic surface rather than the curvature of a line across the surface in some direction
wbt.total_curvature(
    'porce.tif', 
    'porce_curvature_total.tif', 
    log=False, 
    zfactor=None, 
)
```

```python
# Roughness
wbt.average_normal_vector_angular_deviation(
    'porce.tif', 
    'porce_roughness.tif', 
    filter=11, #Size of the filter kernel
)
```

```python
# roughness index calculates the root-mean-square-deviation (RMSD) for each grid cell in a digital elevation model (DEM), calculating the residuals (i.e. elevation differences) between a grid cell and its eight neighbours.
wbt.ruggedness_index(
    'porce.tif', 
    'porce_roughness_index.tif', 
)
```

```python
# circular variance of aspect
wbt.circular_variance_of_aspect(
    'porce.tif', 
    'porce_aspect_variance.tif', 
    filter=11,  #Size of the filter kernel
)
```

```python
#difference between the elevation of each grid cell and the mean elevation of the centering local neighbourhood, normalized by standard deviation
wbt.dev_from_mean_elev(
    'nechi.tif', 
    'nechi_dev_elev.tif', 
    filterx=11, 
    filtery=11, 
)
```

```python
# the difference between the elevation of each grid cell and the mean elevation of the centering local neighbourhood
wbt.diff_from_mean_elev(
    'nechi.tif', 
    'nechi_diff_elev.tif', 
    filterx=11, 
    filtery=11, 
)
```

```python
# Edge density
wbt.edge_density(
    'porce.tif', 
    'porce_edge_density.tif', 
    filter=11, 
    norm_diff=5.0, 
    zfactor=None, 
)
```

```python
# find ridges
wbt.find_ridges(
    "porce.tif", 
    "porce_ridges.tif", 
    line_thin=True, 
)
```

```python
#geomorphons landform classification. Jasiewicz et al. (2013) identified 10 common landform types by reclassifying related geomorphons codes.
#1	Flat, 2	Peak (summit), 3	Ridge, 4	Shoulder, 5	Spur (convex), 6	Slope, 7	Hollow (concave), 8	Footslope, 9	Valley, 10	Pit (depression)
wbt.geomorphons(
    "porce.tif", 
    "porce_geomorphon.tif", 
    search=200, 
    threshold=0.0, 
    fdist=0, 
    skip=0, 
    forms=True, 
    residuals=False
)
```

```python
#hypsometrically tinted shaded relief 
wbt.hypsometrically_tinted_hillshade(
    "nechi_30m.tif", 
    "nechi_30m_tinted_hillshade.tif", 
    altitude=45.0, 
    hs_weight=0.5, 
    brightness=0.5, 
    atmospheric=0.0, 
    palette="atlas", 
    reverse=False, 
    zfactor=None, 
    full_mode=False, 
)
```

```python
# hillshade operation (also called shaded relief) on an input digital elevation model (DEM) with multiple sources of illumination
wbt.multidirectional_hillshade(
    "sucio.tif", 
    "sucio_multi_hillshade.tif", 
    altitude=45.0, 
    zfactor=None, 
    full_mode=False, #Optional flag indicating whether to use full 360-degrees of illumination sources
)
```

```python
# The scheme divides a landscape into seven element types, including: 1 convergent footslopes (CFS), 2 divergent footslopes (DFS), 3 convergent shoulders (CSH), 4 divergent shoulders (DSH), 5 convergent backslopes (CBS), 6 divergent backslopes (DBS), and 7 level terrain (L). 
wbt.pennock_landform_class(
    "porce.tif", 
    "porce_pennock.tif", 
    slope=5.0, #Slope threshold value, in degrees (default is 3.0)
    prof=0.1,  #Profile curvature threshold value (default is 0.1)
    plan=0.0,  #Plan curvature threshold value (default is 0.0)
    zfactor=None, 
)
```

```python
#slope-elevation plot
wbt.slope_vs_elevation_plot(
    "nechi_small.tif", 
    "nechi_small_slope_elevation.html", 
    watershed=None, 
)
```

```python
#stream power index
wbt.stream_power_index(
    "porce_d8_sca.tif", 
    "porce_slope.tif", 
    "porce_powerindex.tif", 
    exponent=1.0, 
)
```

```python
# remove single grid cells with no downslope neighbours.
wbt.breach_single_cell_pits(
    "sucio.tif", 
    "sucio_depression_single.tif", 
)
```

```python
# breach depressions
wbt.breach_depressions(
    "porce.tif", 
    "porce_breached.tif", 
    max_depth=None, #Optional maximum breach depth (default is Inf)
    max_length=None, #Optional maximum breach channel length (in grid cells; default is Inf)
    flat_increment=None, #Optional elevation increment applied to flat areas
    fill_pits=False, #Optional flag indicating whether to fill single-cell pits
)
```

```python
#remove pits single grid cells with no downslope neighbours.
wbt.fill_single_cell_pits(
    "sucio_breached.tif", 
    "sucio_breached_fill_single.tif", 
)
```

```python
#fill all of the depressions in a digital elevation model (DEM) and to remove the flat areas
wbt.fill_depressions(
    "sucio_breached_fill_single.tif", 
    "sucio_filled.tif", 
    fix_flats=True, #Optional flag indicating whether flat areas should have a small gradient applied
    flat_increment=None, #Optional elevation increment applied to flat areas
    max_depth=None, #Optional maximum depression depth to fill
)
```

```python
#a flow pointer grid using the simple D8 (O'Callaghan and Mark, 1984) algorithm
wbt.d8_pointer(
    "nechi_small.tif", 
    "nechi_small_D8_pointer.tif", 
    esri_pntr=False, 
)
```

```python
#generate a flow accumulation grid (i.e. catchment area) using the D8 (O'Callaghan and Mark, 1984) algorithm
wbt.d8_flow_accumulation(
    "nechi_small.tif", 
    "nechi_small_d8_cells.tif", 
    out_type="cells", #'cells' (default), 'catchment area', and 'specific contributing area'
    log=False, #Optional flag to request the output be log-transformed
    clip=False, #Optional flag to request clipping the display max by 1%
    pntr=False, #Is the input raster a D8 flow pointer rather than a DEM?
    esri_pntr=False, 
)
```

```python
#to delineate all of the drainage basins contained within a local drainage direction, or flow pointer raster (--d8_pntr), and draining to the edge of the data
wbt.basins(
    "porce_D8_pointer.tif", 
    "porce_basin.tif", 
    esri_pntr=False, 
)
```

```python
#extract, or map, the likely stream cells from an input flow-accumulation image
wbt.extract_streams(
    "nechi_small_d8_cells.tif", 
    "nechi_small_stream.tif", 
    1000.0, 
    zero_background=False, 
)
```

```python
#identify the catchment areas to each link in a user-specified stream network
wbt.subbasins(
    "porce_D8_pointer.tif", 
    "porce_stream.tif", 
    "porce_subbasin.tif", 
    esri_pntr=False, 
)
```

```python
#ssigns a unique identifier to each tributary in a stream network.
wbt.tributary_identifier(
    "porce_D8_pointer.tif", 
    "porce_stream.tif", 
    "porce_tributary.tif", 
    esri_pntr=False, 
    zero_background=False, 
)
```

```python
#delineates the longest flowpaths for a group of subbasins
wbt.longest_flowpath(
    "sucio_filled.tif", 
    "sucio_basin.tif",
    "sucio_longest_flowpath.tif",  
)
```

```python
#divide a landscape into a group of nearly equal-sized watersheds
wbt.isobasins(
    "sucio_filled.tif", 
    "sucio_isobasins_10000.tif", 
    10000, #Target basin size, in grid cells
    connections=False,
)
```

```python
# calculate the distance from each grid cell in a raster to the nearest stream cell, measured along the downslope flowpath
wbt.downslope_distance_to_stream(
    "porce_filled.tif", 
    "porce_stream.tif", 
    "porce_distance_stream.tif", 
    dinf=False, #Use the D-infinity flow algorithm instead of D8?
)
```

```python
#identify the hillslopes associated with a user-specified stream network
wbt.hillslopes(
    "arma_D8.tif", 
    "arma.stream.tif", 
    "arma_hillslopes.tif", 
    esri_pntr=False, 
)
```
