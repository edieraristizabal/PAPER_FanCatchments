```python
import os
import numpy as np
import matplotlib.pyplot as plt
import imageio
```

```python
filenames=[]
for i in range(19,34):
  filename=f"/content/drive/MyDrive/INVESTIGACION/POSDOC/Perfiles/NS/P_NS{i}.png"
  filenames.append(filename)
# Build GIF
with imageio.get_writer('/content/drive/MyDrive/INVESTIGACION/POSDOC/Perfiles/NS/ProfilesNS19_34.gif', mode='I', duration=0.5) as writer:
    for filename in filenames:
        image = imageio.imread(filename)
        writer.append_data(image)
```
