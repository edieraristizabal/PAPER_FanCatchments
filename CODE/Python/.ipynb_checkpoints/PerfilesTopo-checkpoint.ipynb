```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Profile1: Corte SE

```python
perfil=pd.read_csv('/content/drive/MyDrive/INVESTIGACION/POSDOC/Data/Profile1.csv')
```

```python
x=perfil.iloc[:,0]
y=perfil.iloc[:,1]
```

```python
fig,ax=plt.subplots(figsize=(25,4))
ax.plot(x,y,'g','-',linewidth=2)
ax.set(xlim=(min(x),max(x)),ylim=(min(y),max(y)+500))
ax.set(xlabel='Distance',ylabel='Elevation (m)')
ax.set_xticks(np.arange(min(x), max(x), round((max(x)-min(x))/5,-3)))
ax.set_yticks(np.arange(min(y), max(y), round(max(y)/5,-3)))
ax.set_yticklabels(['','1000','2000','3000','4000'])
ax.tick_params(axis='y',direction='in',length=10)
ax.tick_params(axis='x',direction='in',length=5)
#plt.savefig('profile1.png',dpi=300,transparent=True)
```

Profile 2: SE

```python
perfil=pd.read_csv('/content/drive/MyDrive/INVESTIGACION/POSDOC/Data/Profile2.csv')
x=perfil.iloc[:,0]
y=(perfil.iloc[:,1])
Y=[]
for i in y:
  if isinstance(i,str):
    i = float(i.replace(',',''))
    Y.append(i)
  else: 
    Y.append(i)
```

```python
fig,ax=plt.subplots(figsize=(25,4))
ax.plot(x,Y,'g','-',linewidth=1)
ax.set(xlim=(min(x),max(x)),ylim=(min(Y),max(Y)+500))
ax.set(xlabel='Distance',ylabel='Elevation (m)')
ax.set_xticks(np.arange(min(x), max(x), round((max(x)-min(x))/5,-2)))
ax.set_yticks(np.arange(min(Y), max(Y), round(max(Y)/5,-2)))
ax.set_yticklabels(['','1000','2000','3000','4000'])
ax.tick_params(axis='y',direction='in',length=10)
ax.tick_params(axis='x',direction='in',length=5)
#plt.savefig('profile2.png',dpi=300,transparent=True)
```
