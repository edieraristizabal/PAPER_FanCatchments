```python
import pyosp
import matplotlib.pyplot as plt
from pyosp import read_shape
import matplotlib.image as mpimg
from matplotlib.offsetbox import OffsetImage, AnnotationBbox
import numpy as np
```

```python
cw = pyosp.Orig_curv(
    "/content/drive/MyDrive/INVESTIGACION/POSDOC/Data/Vector/cordilleraW.shp",
    "/content/drive/Shareddrives/Superficie_erosion/DEM_Colombia/Dem_Colombia.tif",
    width=10000,
    line_stepsize=1000,
    cross_stepsize=None
)
```

```python
fig, ax = plt.subplots(figsize = (8,4))
cw.profile_plot(ax=ax, edglecolor="black", color="black")
ax.set_xlabel("Distance (km)", fontsize=18, weight='bold')
ax.set_ylabel("Elevation (m)", fontsize=18, weight='bold')
ax.set_title("Western cordillera (W-W')", fontsize=22, weight='bold',)
ax.xaxis.set_tick_params(labelsize=18)
ax.yaxis.set_tick_params(labelsize=18)
ax.set_xlim(0,700000)
ax.set_ylim(0,5500)
ax.set(xticks=ax.get_xticks()[::2], xticklabels=ax.get_xticklabels()[::2]);
ax.set(yticks=ax.get_yticks()[::2], yticklabels=ax.get_yticklabels()[::2]);
labels = [item.get_text() for item in ax.get_xticklabels()]
labels[0] = '0'
labels[1] = '200'
labels[2] = '400'
labels[3] = '600'
ax.set_xticklabels(labels)
ax.set_facecolor('white')
for tick in ax.get_xticklabels():
    tick.set_fontweight('bold')
for tick in ax.get_yticklabels():
    tick.set_fontweight('bold')
fig.tight_layout()
#plt.legend(fontsize=18)
#ax.grid()

#prof=mpimg.imread(f"/content/drive/MyDrive/INVESTIGACION/POSDOC/Figuras/cordilleraW_location.png")
#imagebox=OffsetImage(prof,zoom=0.055)
#ab=AnnotationBbox(imagebox,(655000,4200),frameon=False)
#ax.add_artist(ab)

ax.get_legend().remove()
plt.savefig('/content/drive/MyDrive/INVESTIGACION/POSDOC/Figuras/cordilleraW.png',dpi=500,
                transparent=False,
                edgecolor="white",
                facecolor=[1,1,1,0])
```

```python
import pandas as pd
difCM=pd.read_excel("../Data/DiferenciasDrenajes.xlsx", sheet_name="CaucavsMagd", engine='openpyxl')
difCP=pd.read_excel("../Data/DiferenciasDrenajes.xlsx", sheet_name="CaucavsPorce", engine='openpyxl')
difMP=pd.read_excel("../Data/DiferenciasDrenajes.xlsx", sheet_name="MagdvsPorce", engine='openpyxl')
difCM.head()
```

```python
fig, ax = plt.subplots(figsize = (15,5))
ax.set_xlabel("E-W Profiles", fontsize=18)
ax.set_ylabel("Elevation difference (m)", fontsize=18)
plt.plot(difCM.x, difCM.Dif_Cotas, c="blue", label='Magdalena-Cauca')
plt.plot(difCP.x, difCP.Dif_Cotas, c="green", label='Cauca-Porce')
plt.plot(difMP.x, difMP.Dif_Cotas, c="red", label='Magdalena-Porce')
plt.xlim(min(difCM.x),max(difCM.x))
ax.text("P09",570,"Arma r.",fontsize=14,rotation="vertical")
ax.text("P15",450,"Ancón sur",fontsize=14,rotation="vertical")
ax.text("P19",1070,"Bello",fontsize=14,rotation="vertical")
ax.text("P22",850,"La Montera",fontsize=14,rotation="vertical")
ax.text("P27",180,"Espíritu Santo",fontsize=14,rotation="vertical")
ax.legend()
ax.grid()
plt.savefig('../Figuras/riverElevationDifferences.png',dpi=400,transparent=True)
```
