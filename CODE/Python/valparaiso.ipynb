```python
#to import data for chi and longitudinal profiles
data= gpd.read_file("/valparaiso2.shp")
data.head()
```

```python
#To select basin (when there are several)
data=data[data["basin_key"].isin([0])]
```

```python
fig, ax = plt.subplots(figsize = (15,5))
#to print the entire number of drainage
ax.scatter(data.flow_dista/1000, data.elevation, s=0.1)


ax.set_xlabel('Flow distance ($km$)', fontsize=18)
ax.set_ylabel('Elevation ($m.a.s.l.$)', fontsize=18)
#ax.set_xlim(0,450)
#ax.set_ylim(0,3500)
plt.savefig('/perfil.png',dpi=400,transparent=True)
```

```python
fig, ax = plt.subplots(figsize = (15,5))
ax.scatter(data.chi, data.elevation, s=0.1)

ax.set_xlabel('$\chi$ distance', fontsize=18)
ax.set_ylabel('Elevation (m)', fontsize=18)
#ax.set_xlim(0,15)
#ax.set_ylim(0,3500)
#plt.savefig('/content/drive/MyDrive/INVESTIGACION/POSDOC/Nechi/Images/nechi_chi_elevation_12m.png',dpi=400,transparent=True)
```
