```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
perfil=pd.read_excel('/content/drive/MyDrive/INVESTIGACION/SEGUIMIENTO ESTUDIANTES/PAE/Manuela Renteria/datos.xlsx', sheet_name="Q. Altavista")
```

```python
perfil.head(2)
```

```python
x=perfil.iloc[:,0]
y=perfil.iloc[:,1]
z=perfil.iloc[:,2]
```

```python
Xmax=np.max(x)
Xmin=np.min(x)
Ymin=np.min(y)
Ymax=np.max(y)
```

```python
fig,ax=plt.subplots(figsize=(14,4))
ax.plot(x,y,'b','-',linewidth=1)
ax.set(xlim=(Xmin,Xmax),ylim=(Ymin,Ymax))
ax.set(xlabel='Distance (km)',ylabel='Elevation (m)')
plt.xticks(np.arange(min(x), max(x), round((max(x)-min(x))/5,-3)))
plt.yticks(np.arange(min(y), max(y), round((max(y)-min(y))/5,-2)))
ax.tick_params(axis='y',direction='in',length=10)
ax.tick_params(axis='x',direction='in',length=5)
#plt.savefig('riosucio.png',dpi=300,transparent=True)
```
