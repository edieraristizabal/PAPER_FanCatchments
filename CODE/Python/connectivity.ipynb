# connectivity

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sbn
import geopandas as gpd
from pysal.lib import weights
from matplotlib_scalebar.scalebar import ScaleBar
import geodatasets
from geo_northarrow import add_north_arrow
```

```python
gdf=gpd.read_file("G:/My Drive/INVESTIGACION/POSDOC/Data/Vector/catchments_OK.gpkg")
```

```python
gdf.plot();
```

```python
gdf.head(1)
```

```python
w_queen = weights.contiguity.Rook.from_dataframe(gdf);
```

```python
w_queen.nonzero
```

```python
f, ax = plt.subplots(figsize=(9, 9))
gdf.plot(facecolor="none", edgecolor="gray",ax=ax)
w_queen.plot(gdf,edge_kws=dict(linewidth=0.7,color="red"),node_kws=dict(marker="."),ax=ax,)
add_north_arrow(ax, scale=.75, xlim_pos=.2, ylim_pos=.965, color='#000', text_scaler=4, text_yT=-1.25)
scalebar = ScaleBar(1,"m",location="lower right",scale_loc="top",length_fraction=0.25)
ax.add_artist(scalebar);
ax.set_axis_off()
ax.set_axis_off()
f.savefig('G:/My Drive/INVESTIGACION/POSDOC/Figuras/w_queen.png', dpi=500, bbox_inches='tight')
```

```python
ax = gdf.plot(color='k', figsize=(9, 9))
gdf.loc[w_queen.islands, :].plot(color='red', ax=ax);
```
